Create a function called price_category that returns a string based onthe fare paid. For example, 0-75 dollars returns "Budget", 75-150 dollars returns "Medium" and 150+ dollars returns "Expensive". Use this function to create a new column called 'price.category'.

```python
def price_category(fare):
    if fare >= 150:
        return "Expensive"
    elif fare >= 75:
        return "Medium"
    else:
        return "Budget"
titanic['price.category'] = titanic['fare'].apply(price_category)
titanic.head(10)
```

Create a function called label_person that returns a summary tag given the person's gender and age. For example, male gender and under 40 years old returns "Younger man", male gender and over 40 years old returns "Older man", etc. Use this function to create a new column called 'person.label'.

```python
def label_person(row):
    if row['sex'] == "male":
        if row['age'] < 40:
            return "Younger man"
        elif row['age'] >= 40:
            return "Older man"
    elif row['sex'] == "female":
        if row['age'] < 40:
            return "Younger woman"
        elif row['age'] >= 40:
            return "Older woman"
titanic['person.label'] = titanic.apply(label_person, axis=1)
titanic.head(20)
```

Find the average survival rate, grouped by person.label and then by price.category. Hint: Use mean of survived column, since 0 means didn't survive and 1 means they did.

```python
avgservy = titanic.groupby(['person.label','price.category'])['survived'].mean().unstack()
avgservy
```
